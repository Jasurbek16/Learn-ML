```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
X = np.random.randn(150, 2)
y = np.array([0]*50 + [1]*50 + [2]*50) # Three classes
```

```python
plt.scatter(X[y==0, 0], X[y==0, 1], label='Class A') # scatter plot for class A; choosing feature columns of rows of a class during scatter plot
plt.scatter(X[y==1, 0], X[y==1, 1], label='Class B')
plt.scatter(X[y==2, 0], X[y==2, 1], label='Class C')
plt.legend()
plt.title('Multi-class Data')
plt.show()
```

```python
def softmax(z):
    return np.exp(z) / np.sum(np.exp(z), axis=1, keepdims=True) # axis=1 - one sum per sample; keepdims=True - keep the dimensions of the output same as input
```

```python
z = np.array([[1.0, 2.0, 3.0]])
print(softmax(z))
print(np.sum(softmax(z)))
```
